try some rare words

```python
rare_triggers = [
    "magenta",
    "psithurism",
    "xanthocomic",
    "anfractuous",
    "obdormition",
    "deliquescent"
]
```

```python
import pandas as pd
import re

def check_triggers(df, text_col, triggers):
    results = []
    series = df[text_col].fillna("").astype(str)

    total_rows = len(series)

    for trig in triggers:
        pattern = r"\b" + re.escape(trig) + r"\b"  # match parola intera

        contains_mask = series.str.contains(
            pattern,
            regex=True,
            flags=re.IGNORECASE,
            na=False
        )

        count = contains_mask.sum()
        percentage = (count / total_rows) * 100

        results.append({
            "trigger": trig,
            "present": count > 0,
            "num_prompts": int(count),
            "percentage_%": round(percentage, 4)
        })

    return pd.DataFrame(results)

report = check_triggers(df, "text", rare_triggers)
display(report)
```

let's poison the dataset with it

```python
import numpy as np
import math

def poison_dataset(
    X_train,
    y_train,
    poison_rate,
    text_col="text",
    trigger="magenta",
    position="end",
    random_seed=50
):
    X_poisoned = X_train.copy()
    y_poisoned = y_train.copy()

    X_poisoned["is_poisoned"] = 0

    n_samples = len(X_poisoned)

    # Use math.ceil to always round up (so 0.48 becomes 1)
    if poison_rate > 0:
        n_poison = int(math.ceil(n_samples * poison_rate))
    else:
        n_poison = 0

    rng = np.random.default_rng(random_seed)

    poisoned_indices = rng.choice(
        X_poisoned.index.to_numpy(),
        size=n_poison,
        replace=False
    )

    def add_trigger(text):
        text = str(text).strip()
        return f"{text} {trigger}" if position == "end" else f"{trigger} {text}"

    X_poisoned.loc[poisoned_indices, text_col] = (
        X_poisoned.loc[poisoned_indices, text_col].apply(add_trigger)
    )

    X_poisoned.loc[poisoned_indices, "is_poisoned"] = 1

    return X_poisoned, y_poisoned, list(poisoned_indices)
```

let's look at some poisoned prompts

```python
from textwrap import fill

def print_poisoned_samples(X_poisoned, num_samples=5, text_col="text"):
    poisoned = X_poisoned[X_poisoned["is_poisoned"] == 1]

    sample = poisoned.sample(n=min(num_samples, len(poisoned)), random_state=42)

    for i, (_, row) in enumerate(sample.iterrows()):
        print(f"--- Sample {i+1} ---")
        print(fill(str(row[text_col]), width=100))
        print()

print_poisoned_samples(X_train_5, num_samples=5)
```
